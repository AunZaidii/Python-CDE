```python
class student:
    def __init__(self,name,age,course):
        self.name = name
        self.age = age
        self.course = course
        self.grades = []

    def introduce(self):
        print(f"Hi, I am {self.name}. My age is {self.age} and i am enrolled in {self.course}")

    def add_grade(self, grade):
        self.grades.append(grade)

    def avg_grade(self):
        if not self.grades:
            return 0
        return sum(self.grades) / len(self.grades)
        
ali = student("Ali",20,"Data Engineering")
print(ali.name)
print(ali.introduce())
ali.add_grade(90)
ali.add_grade(50)
ali.add_grade(80)

print(f"Hi! My name is {ali.name} and my average grade is {ali.avg_grade()}")
```

```python
#ENCAPSULATION

class bankaccount:
    def __init__(self,owner,balance):
        self.owner = owner                 #public with no _
        self._balance = balance            #protected with one _
        self.__pin = "0000"                #private with two__

    def deposit(self,amount):
        self._balance += amount

    def get_balance(self):
        return self._balance

    def get_pin(self):
        print(f"your pin is {self.__pin}")
    
    def change_pin(self,old_pin, new_pin):
        if old_pin == self.__pin:
            self.__pin = new_pin
            return "PIN Updated"
        return "Incorrect PIN"
        
acc_bilal = bankaccount("Bilal",1000)
print(acc_bilal.get_balance())
acc_bilal.deposit(500)
print(acc_bilal.get_balance())
acc_bilal.change_pin("0000","1111")
print(acc_bilal.get_pin())
# print(acc_bilal.__pin)    will not work
```

```python
#INHERITANCE

class person:
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"Hi I am {self.name} and I am {self.age} year old")

class student(person):
    def __init__ (self,name,age,course):
        super().__init__(name,age)
        self.course = course

    def introduce_student(self):
        
        print(f"Hi I am {self.name} and I am {self.age} year old and I am studying {self.course}")
    
class teacher(person):
    def __init__ (self,name,age,teach):
        super().__init__(name,age)
        self.teach = teach


    def introduce_teacher(self):
        print(f"Hi I am {self.name} and I am {self.age} year old and I am teaching {self.teach}")

class teacher_assistant(teacher):
    def __init__ (self,name,age,teach,research):
        super().__init__(name,age,teach)
        self.research = research


    def introduce_teacher_assistant(self):
        print(f"Hi I am {self.name} and I am {self.age} year old and I am teaching {self.teach} and I am researching on {self.research}")

raza = person("Raza",29)
ali = student("Ali",20,"OOP")
ahmed = teacher("Ahmed",35,"OOP")
zia = teacher_assistant("Zia",25,"OOP","Encapsulation")
raza.introduce()
ali.introduce_student()
ahmed.introduce_teacher()
zia.introduce_teacher_assistant()
```

```python
#POLYMORPHISM

class person:
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"Hi I am {self.name} and I am {self.age} year old")

class student(person):
    def __init__ (self,name,age,course):
        super().__init__(name,age)
        self.course = course

    def introduce(self):
        
        print(f"Hi I am {self.name} and I am {self.age} year old and I am studying {self.course}")
    
class teacher(person):
    def __init__ (self,name,age,teach):
        super().__init__(name,age)
        self.teach = teach


    def introduce(self):
        print(f"Hi I am {self.name} and I am {self.age} year old and I am teaching {self.teach}")

class teacher_assistant(teacher):
    def __init__ (self,name,age,teach,research):
        super().__init__(name,age,teach)
        self.research = research


    def introduce(self):
        print(f"Hi I am {self.name} and I am {self.age} year old and I am teaching {self.teach} and I am researching on {self.research}")

raza = person("Raza",29)
ali = student("Ali",20,"OOP")
ahmed = teacher("Ahmed",35,"OOP")
zia = teacher_assistant("Zia",25,"OOP","Encapsulation")
raza.introduce()
ali.introduce()
ahmed.introduce()
zia.introduce()
```

```python
#ABSTRACTION

from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
        
    def area(self):
        return (self.radius**2) * 22.7

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

circle = Circle(10)
rectangle = Rectangle(2,4)

circle.area()
rectangle.area()
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Name: {self.name}, Age: {self.age}"


class Student(Person):
    def __init__(self, name, age):
        super().__init__(name, age)
        self.grades = []

    def add_grade(self, grade):
        self.grades.append(grade)

    def avg_grade(self):
        return sum(self.grades) / len(self.grades)

    def summary(self):
        print(self)
        print("Grades:", self.grades)
        print("Average Grade:", self.avg_grade())


# Mini Classroom
student1 = Student("Bilal", 20)
student1.add_grade(80)
student1.add_grade(90)
student1.add_grade(85)

student1.summary()
```
