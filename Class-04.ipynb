```python
import csv
import numpy as np

quantities, prices = [], []
with open(r"C:\Users\Aun Zaidi\Downloads\brewbuzz_raw_orders.csv", newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        try:
            q = int(row["quantity"])
            p = float(row["price"])
            if q > 0 and p > 0:
                quantities.append(q)
                prices.append(p)
        except ValueError:
            continue
qty_arr = np.array(quantities)
price_arr = np.array(prices)
```
