```python
import csv
with open(r"C:\Users\Aun Zaidi\Downloads\supermarket_customers.csv","r",encoding="utf-8") as f_in, \
     open(r"C:\Users\Aun Zaidi\Downloads\airport_orders.csv.txt", "w", newline="", encoding="utf-8") as f_out:
    reader = csv.DictReader(f_in)
    writer = csv.DictWriter(f_out, fieldnames=reader.fieldnames)
    writer.writeheader()
    for row in reader:
        if row["Gender"] == "Male":
            writer.writerow(row)

with open(r"C:\Users\Aun Zaidi\Downloads\airport_orders.csv.txt", encoding="utf-8") as f:
    print(sum(1 for _ in f) - 1, "rows written (excluding header)")
```

```python
import csv
with open(r"C:\Users\Aun Zaidi\Downloads\brewbuzz_raw_orders.csv","r",encoding="utf-8") as f_in, \
     open(r"C:\Users\Aun Zaidi\Downloads\orders.csv.txt", "w", newline="", encoding="utf-8") as f_out:
    reader = csv.DictReader(f_in)
    writer = csv.DictWriter(f_out, fieldnames=["order_id","order_date","store_name","customer_email","order_total"])
    writer.writeheader()
    for row in reader:
        if  "2025-01-01" <= row["order_date"] <= "2025-12-01":
            writer.writerow({"order_id": row["order_id"], "order_date": row["order_date"], "store_name": row["store_name"], "customer_email": row["customer_email"], "order_total": row["order_total"] })

with open(r"C:\Users\Aun Zaidi\Downloads\orders.csv.txt", encoding="utf-8") as f:
    print(sum(1 for _ in f) - 1, "rows written (excluding header)")
```
